# 05逻辑回归：交叉熵与梯度
## 为什么不用MSE均方误差？
$$ L_{MSE} = (y-\sigma(z))^2 $$
缺点：
* 损失函数非凸，存在很多局部极小，梯度下降容易跑不到全局最优,找不到全局最好的 w,b
* sigmoid 饱和区域梯度很小，参数更新极慢,训练极其缓慢
> 👉 分类任务使用：交叉熵损失（Log Loss）

## 单样本交叉熵损失
- 标签 $y$ 只有两种取值：0 或 1。
- $\hat{p}$ 为模型预测的"正类（y=1）概率"，取值范围 $(0, 1)$。

$$
L(\hat{p}, y) =
\begin{cases}
-\log(\hat{p}) & y = 1 \\
-\log(1 - \hat{p}) & y = 0
\end{cases}
$$

统一写法：
$$ L = -\Big[y\cdot \log(\hat p)+(1-y)\cdot \log(1-\hat p)\Big] $$
理解：
* $y=1$：希望 $\hat p$ 靠近 1，预测错误时损失急剧变大
* $y=0$：希望 $\hat p$ 靠近 0，预测错误时损失急剧变大
> ⚠️ 代码中必须增加极小偏移量 eps=1e-8，避免 $\log(0)$ 无穷报
**为什么等价**：
- 当 $y = 1$ 时，$(1-y)=0$，第二项消失，剩 $-\log(\hat{p})$；
- 当 $y = 0$ 时，$y=0$，第一项消失，剩 $-\log(1-\hat{p})$。
> 即用一个式子自动"选择"对应分支，便于向量化实现。


## 全部样本代价函数
一共有 $n$ 个样本，取平均损失：
$$ J(\boldsymbol w,b)= -\frac1n\sum_{i=1}^n \Big[y_i\cdot \log(\hat p_i)+(1-y_i)\cdot \log(1-\hat p_i)\Big] $$
✅ 该函数是凸函数，合适的学习率下梯度下降一定可以找到全局最优。

## 梯度推导
Sigmoid 导数性质：
$$ \sigma'(z)=\sigma(z)\cdot(1-\sigma(z)) $$
链式求导化简得到最终梯度：
$$ \frac{\partial J}{\partial \boldsymbol w} = \frac1n \boldsymbol X^T (\hat{\boldsymbol p}-\boldsymbol y) $$
$$ \frac{\partial J}{\partial b} = \frac1n \sum_{i=1}^n (\hat p_i - y_i) $$
> ✨ 对比线性回归梯度公式，形式几乎一模一样，区别是 $\hat p$ 是 sigmoid 输出的概率

## 梯度下降更新
梯度方向是损失上升方向，参数向反方向更新：
$$ \boldsymbol w = \boldsymbol w - \alpha \cdot \frac{\partial J}{\partial \boldsymbol w} $$
$$ b = b - \alpha \cdot \frac{\partial J}{\partial b} $$
$\alpha$：学习率
* $\alpha$ 太大：损失震荡，不收敛
* $\alpha$ 太小：收敛速度很慢
