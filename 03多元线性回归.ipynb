# 03多元线性回归
## 什么是多元线性回归
简单线性回归：**只有1个特征**，拟合直线
$$\hat y = w_0 + w_1 x_1$$

多元线性回归：**≥2个特征**，拟合的不再是直线，是**超平面**
$$
\boldsymbol{\hat y = w_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + \dots + w_n x_n}
$$
- $x_1,x_2...x_n$：不同特征（自变量），比如：房屋面积、房间数、楼层
- $w_1,w_2...w_n$：各个特征对应的权重（系数），代表该特征对y的影响大小
- $w_0$：截距

矩阵写法（推荐记住，sklearn底层、闭式解都用矩阵）：
$$
\hat y = \boldsymbol{XW}
$$
$\boldsymbol{X}$ 是增加了一列全1的设计矩阵，$\boldsymbol{W}=[w_0,w_1,w_2,...]^T$

> 例子：房价预测
> $y$ = 房价
> $x_1$ = 面积，$x_2$ = 房间数量，$x_3$ = 到地铁距离
> $\hat y = w_0 + w_1\cdot面积 + w_2\cdot房间数 + w_3\cdot地铁距离$

## 损失函数
依旧使用MSE均方误差：
$$
J(w_0,w_1...w_n) = \frac{1}{n}\sum_{i=1}^n (y_i-\hat y_i)^2
$$
目标：找到一组 $w_0,w_1...w_n$，让MSE最小。

两种求解方式：
1. **闭式解（矩阵最小二乘）**：sklearn `LinearRegression` 默认用这个，直接一步算出最优参数
2. **梯度下降**：迭代更新所有参数，适合特征极多、大数据场景

## 评价指标
和简单线性回归完全一样：$R^2$ 决定系数
$$
R^2 = 1-\frac{\sum(y_i-\hat y_i)^2}{\sum(y_i-\bar y)^2}
$$
- $R^2 \le 1$
- 越接近1，说明多个特征组合起来对y的解释能力越强
- $R^2<0$：模型不如直接用均值预测

> ✅ 重要区分：多元里有个坑——**R²会随着特征增加虚高**，后面会引出「调整R²」

```python
# 自动生成house.csv
import pandas as pd
data = {
    "Area":[60,75,88,102,50,120,95,110,68,82],
    "Room":[2,3,3,4,2,4,3,4,2,3],
    "Price":[120,150,180,210,95,260,200,240,140,170]
}
df = pd.DataFrame(data)
df.to_csv("./house.csv",index=False,encoding="utf-8")
```

```python
# sklearn实现
# sklearn 多元线性回归（修复警告 + 中文显示）
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import warnings

warnings.filterwarnings("ignore")
# 解决matplotlib中文乱码
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# 数据集：房价，特征：面积、房间数；标签：房价
df = pd.read_csv("./house.csv")
# 多个特征，直接传入多列，天然二维，不需要额外reshape
X = df[["Area", "Room"]]
y = df["Price"]

# 划分训练、测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# 训练模型，和简单线性回归代码完全一样！
reg = LinearRegression()
reg.fit(X_train, y_train)

# 获取参数
w_list = reg.coef_   # 权重数组 [w1,w2]，对应Area、Room
w0 = reg.intercept_
print(f"截距 w0 = {w0:.2f}")
print(f"面积权重 w1 = {w_list[0]:.2f}")
print(f"房间数权重 w2 = {w_list[1]:.2f}")
print(f"公式：y = {w0:.2f} + {w_list[0]:.2f}*面积 + {w_list[1]:.2f}*房间数")
# 预测评估
y_pred = reg.predict(X_test)
print(f"R² = {r2_score(y_test, y_pred):.3f}")
```
